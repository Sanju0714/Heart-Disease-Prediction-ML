### Comparison of Missing Value Imputation Techniques

```python
### Handling Missing Values - Multiple Imputation Techniques

from sklearn.impute import SimpleImputer, KNNImputer

# Keep original dataset
df_original = df.copy()

# Mean Imputation
mean_imputer = SimpleImputer(strategy='mean')
df_mean = pd.DataFrame(
    mean_imputer.fit_transform(df_original),
    columns=df_original.columns
)

# Median Imputation
median_imputer = SimpleImputer(strategy='median')
df_median = pd.DataFrame(
    median_imputer.fit_transform(df_original),
    columns=df_original.columns
)

# KNN Imputation
knn_imputer = KNNImputer(n_neighbors=5)
df_knn = pd.DataFrame(
    knn_imputer.fit_transform(df_original),
    columns=df_original.columns
)

print("Missing values after Mean Imputation:")
print(df_mean.isnull().sum())

print("\nMissing values after Median Imputation:")
print(df_median.isnull().sum())

print("\nMissing values after KNN Imputation:")
print(df_knn.isnull().sum())
```

### Feature Selection

```python
## Separate Features and Target
X = df.drop("TenYearCHD", axis=1)
y = df["TenYearCHD"]
```

```python
## Display Features
X.head()
```

```python
## Display Target
y.head()
```

```python
## Check Feature Names
X.columns
```

```python
## Import Feature Selection Library
from sklearn.feature_selection import SelectKBest
from sklearn.feature_selection import f_classif

selector = SelectKBest(score_func=f_classif, k='all')

selector.fit(X, y)

feature_scores = pd.DataFrame({
    "Feature": X.columns,
    "Score": selector.scores_
})

feature_scores.sort_values(by="Score", ascending=False)
```

```python
## Visualize Feature Importance
plt.figure(figsize=(10,6))

sns.barplot(
    data=feature_scores.sort_values(by="Score", ascending=False),
    x="Score",
    y="Feature"
)

plt.title("Feature Importance using ANOVA F-Test")

plt.show()
```

```python
## Select Top Features

selector = SelectKBest(score_func=f_classif, k=8)

X_selected = selector.fit_transform(X, y)

selected_features = X.columns[selector.get_support()]

print(selected_features)
```

### Train-Test Split

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,          # or X_selected if you selected top features
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)
```

### Feature Scaling

```python
## Import StandardScaler
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)


import joblib
joblib.dump(scaler, "scaler.pkl")
```

### Model Building

```python
## Import Required Libraries
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)
```

### Logistic Regression

```python
##Train the Model
lr_model = LogisticRegression(max_iter=1000, random_state=42)

lr_model.fit(X_train, y_train)
```

```python
##Make Predictions
lr_pred = lr_model.predict(X_test)
```

```python
## Evaluate Logistic Regression
print("Logistic Regression")

print("Accuracy :", accuracy_score(y_test, lr_pred))
print("Precision:", precision_score(y_test, lr_pred))
print("Recall   :", recall_score(y_test, lr_pred))
print("F1 Score :", f1_score(y_test, lr_pred))
```

```python
##Confusion Matrix
cm = confusion_matrix(y_test, lr_pred)

plt.figure(figsize=(5,4))

sns.heatmap(cm,
            annot=True,
            fmt="d",
            cmap="Blues")

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Logistic Regression Confusion Matrix")

plt.show()
```

```python
##Classification Report
print(classification_report(y_test, lr_pred))
```

### Decision Tree

```python
##Train Model
dt_model = DecisionTreeClassifier(random_state=42)

dt_model.fit(X_train, y_train)
```

```python
## Prediction
dt_pred = dt_model.predict(X_test)
```

```python
##Evaluation
print("Decision Tree")

print("Accuracy :", accuracy_score(y_test, dt_pred))
print("Precision:", precision_score(y_test, dt_pred))
print("Recall   :", recall_score(y_test, dt_pred))
print("F1 Score :", f1_score(y_test, dt_pred))
```

### Random Forest

```python
## Train Model
rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

rf_model.fit(X_train, y_train)
```

```python
##Prediction
rf_pred = rf_model.predict(X_test)
```

```python
##Evaluation
print("Random Forest")

print("Accuracy :", accuracy_score(y_test, rf_pred))
print("Precision:", precision_score(y_test, rf_pred))
print("Recall   :", recall_score(y_test, rf_pred))
print("F1 Score :", f1_score(y_test, rf_pred))
```

### Compare All Models

```python
results = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Decision Tree",
        "Random Forest"
    ],
    "Accuracy": [
        accuracy_score(y_test, lr_pred),
        accuracy_score(y_test, dt_pred),
        accuracy_score(y_test, rf_pred)
    ],
    "Precision": [
        precision_score(y_test, lr_pred),
        precision_score(y_test, dt_pred),
        precision_score(y_test, rf_pred)
    ],
    "Recall": [
        recall_score(y_test, lr_pred),
        recall_score(y_test, dt_pred),
        recall_score(y_test, rf_pred)
    ],
    "F1 Score": [
        f1_score(y_test, lr_pred),
        f1_score(y_test, dt_pred),
        f1_score(y_test, rf_pred)
    ]
})

results
```
