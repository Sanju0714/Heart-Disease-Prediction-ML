### Comparison of Missing Value Imputation Techniques

```python
### Handling Missing Values - Multiple Imputation Techniques

from sklearn.impute import SimpleImputer, KNNImputer

# Keep original dataset
df_original = df.copy()

# Mean Imputation
mean_imputer = SimpleImputer(strategy='mean')
df_mean = pd.DataFrame(
    mean_imputer.fit_transform(df_original),
    columns=df_original.columns
)

# Median Imputation
median_imputer = SimpleImputer(strategy='median')
df_median = pd.DataFrame(
    median_imputer.fit_transform(df_original),
    columns=df_original.columns
)

# KNN Imputation
knn_imputer = KNNImputer(n_neighbors=5)
df_knn = pd.DataFrame(
    knn_imputer.fit_transform(df_original),
    columns=df_original.columns
)

print("Missing values after Mean Imputation:")
print(df_mean.isnull().sum())

print("\nMissing values after Median Imputation:")
print(df_median.isnull().sum())

print("\nMissing values after KNN Imputation:")
print(df_knn.isnull().sum())
```

### Data Preprocessing

```python
### Separate Features (X) and Target (y)

# Features (Independent Variables)
X = df.drop("TenYearCHD", axis=1)

# Target (Dependent Variable)
y = df["TenYearCHD"]
```

```python
### Check Shapes
print("Features Shape:", X.shape)
print("Target Shape:", y.shape)
```

```python
### Train-Test Split

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
### Verify Split
print("X_train:", X_train.shape)
print("X_test :", X_test.shape)

print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

### Feature Scaling

```python

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)

X_test = scaler.transform(X_test)
```

fit_transform() calculates the mean and standard deviation only from the training data and scales it.
transform() uses the same values to scale the test data.

### Model Building

### Logistic Regression

```python
### Train the Model
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression(random_state=42)

lr.fit(X_train, y_train);
```

```python
### Make Predictions
y_pred_lr = lr.predict(X_test)
```

```python
### Evaluate the Model
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

print("Accuracy:", accuracy_score(y_test, y_pred_lr))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_lr))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_lr))
```

### Decision Tree

```python
### Train the Model
from sklearn.tree import DecisionTreeClassifier

dt = DecisionTreeClassifier(random_state=42)

dt.fit(X_train, y_train)
```

```python
### Make Predictions
y_pred_dt = dt.predict(X_test)
```

```python
### Evaluate the Model
print("Accuracy:", accuracy_score(y_test, y_pred_dt))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_dt))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_dt))
```

### Random forest

```python
### Train the Model
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(random_state=42)

rf.fit(X_train, y_train)
```

```python
### Make Predictions
y_pred_rf = rf.predict(X_test)
```

```python
### Evaluate the Model
print("Accuracy:", accuracy_score(y_test, y_pred_rf))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_rf))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_rf))
```

```python
### Evaluate the Model
print("Accuracy:", accuracy_score(y_test, y_pred_rf))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_rf))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_rf))
```

Logistic Regression: A simple baseline classification model that's easy to interpret.
Decision Tree: Captures non-linear relationships and is easy to visualize.
Random Forest: An ensemble of decision trees that generally provides better accuracy and reduces overfitting.
